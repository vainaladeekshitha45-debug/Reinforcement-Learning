```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Initialize tables
N = {'A': 0, 'B': 0, 'C': 0}       # Number of times played
R = {'A': [], 'B': [], 'C': []}    # Rewards received
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Estimated Q-values

epsilon = 0.2
num_plays = 10

print("Modified ε-Greedy Strategy for Multi-Armed Bandit")
print("-" * 55)

for play in range(1, num_plays + 1):

    print(f"\nPlay {play}")

    # Find machine with highest Q-value
    highest_q = max(Q.values())

    # If all Q-values are equal, choose randomly
    best_machines = [m for m in machines if Q[m] == highest_q]
    best_machine = random.choice(best_machines)

    # Generate random number
    r = random.random()

    # Part C - Modified ε-Greedy Decision
    if r < epsilon:
        # Exploration: choose any machine except best machine
        other_machines = [m for m in machines if m != best_machine]
        selected_machine = random.choice(other_machines)
        print("Mode: Exploration")
    else:
        # Exploitation: choose machine with highest Q-value
        selected_machine = best_machine
        print("Mode: Exploitation")

    print("Selected Machine:", selected_machine)

    # Part D - Reward Collection
    reward = float(input("Enter reward: "))

    # Store reward
    R[selected_machine].append(reward)

    # Update number of plays
    N[selected_machine] += 1

    # Part E - Incremental Sample Average Q-value Update
    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    print("Updated Q-value:", round(Q[selected_machine], 2))


# Part F - Display Results
print("\n" + "=" * 55)
print("FINAL RESULTS")
print("=" * 55)

total_reward = 0

print(f"{'Machine':<10}{'N':<10}{'Rewards':<25}{'Q-value'}")

for machine in machines:
    total_reward += sum(R[machine])

    print(
        f"{machine:<10}"
        f"{N[machine]:<10}"
        f"{str(R[machine]):<25}"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward Received =", total_reward)
```
