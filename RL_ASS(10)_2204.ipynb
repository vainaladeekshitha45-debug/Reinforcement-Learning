```python
# Practical 10
# Bellman Optimality and Value Iteration
# Deterministic Grid World

# Parameters
gamma = 0.9
theta = 0.000001

# States and actions
states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

terminal_states = ['Wall', 'G']

# Reward and transition mapping
# Format: (Next State, Reward)

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('Wall', -1)
    },

    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 10)
    }
}


# Function to get state value
def get_value(state, values):

    if state in terminal_states:
        return 0

    return values[state]


# Part 1: Inspect MDP
print("GRID WORLD MDP")
print("----------------")

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward: -1")
print("Wall Reward: -1")
print("Goal Reward: +10")


# Part 2: Verify transitions
print("\nTRANSITION VERIFICATION")
print("-----------------------")

for state in states:

    for action in actions:

        next_state, reward = transitions[state][action]

        print(state, "+", action,
              "->", next_state,
              "Reward =", reward)


# Part 3: Initialize state values
V = {
    'S0': 0.0,
    'S1': 0.0,
    'Wall': 0.0,
    'G': 0.0
}

print("\nINITIAL STATE VALUES")
print("--------------------")

print(V)


# Part 4: Value Iteration
print("\nVALUE ITERATION")
print("---------------")

iteration = 0

print("Iteration | V(S0) | V(S1) | Delta")

while True:

    # Synchronous update:
    # Use only previous iteration values
    old_V = V.copy()

    new_V = old_V.copy()

    for state in states:

        q_values = []

        for action in actions:

            next_state, reward = transitions[state][action]

            q = reward + gamma * get_value(
                next_state, old_V
            )

            q_values.append(q)

        new_V[state] = max(q_values)

    # Calculate maximum change
    delta = max(
        abs(new_V['S0'] - old_V['S0']),
        abs(new_V['S1'] - old_V['S1'])
    )

    V = new_V

    iteration += 1

    print(
        iteration, "|",
        round(V['S0'], 6), "|",
        round(V['S1'], 6), "|",
        round(delta, 6)
    )

    # Check convergence
    if delta < theta:
        break


# Part 5: Display optimal state values
print("\nOPTIMAL STATE VALUES")
print("--------------------")

for state in states:
    print("V*(", state, ") =", round(V[state], 2))


# Part 6: Calculate action values
print("\nACTION VALUES")
print("-------------")

Q = {}

for state in states:

    Q[state] = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        Q[state][action] = (
            reward + gamma * get_value(next_state, V)
        )

        print(
            "Q*(", state, ",", action, ") =",
            round(Q[state][action], 2)
        )


# Part 7: Extract optimal policy
print("\nOPTIMAL POLICY")
print("--------------")

policy = {}

for state in states:

    max_q = max(Q[state].values())

    best_actions = [
        action
        for action in actions
        if abs(Q[state][action] - max_q) < 1e-9
    ]

    policy[state] = best_actions

    print(state, ":", best_actions)


# Part 8: Bellman Optimality verification
print("\nBELLMAN OPTIMALITY VERIFICATION")
print("--------------------------------")

for state in states:

    max_q = max(Q[state].values())

    print(
        state,
        "V* =", round(V[state], 2),
        "Max Q* =", round(max_q, 2),
        "Verified =", abs(V[state] - max_q) < 1e-9
    )


# Part 9: Final result
print("\nFINAL RESULT")
print("------------")

print("Total Iterations:", iteration)
print("Optimal V(S0):", round(V['S0'], 2))
print("Optimal V(S1):", round(V['S1'], 2))
print("Optimal Policy:", policy)
```
