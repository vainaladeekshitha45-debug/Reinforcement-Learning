```python
# Greedy Multi-Armed Bandit
# Arithmetic Mean Method

n = int(input("Enter number of machines: "))

machines = {}

for i in range(n):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards separated by spaces: ").split()))

    machines[name] = rewards

print("\n--- Estimated Q(a) Values ---")

q_values = {}

for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[name] = q

    print(f"Machine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q:.2f}")

selected_machine = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```

```python
# Greedy Multi-Armed Bandit
# Incremental Update Method

n = int(input("Enter number of machines: "))

machines = []
rewards = {}

# Input machines and rewards
for i in range(n):
    name = input(f"\nEnter name of machine {i + 1}: ")
    reward_list = list(map(float, input("Enter observed rewards separated by spaces: ").split()))

    machines.append(name)
    rewards[name] = reward_list

# Initialize Q and N
Q = {}
N = {}

for machine in machines:
    Q[machine] = 0
    N[machine] = 0

print("\n--- Incremental Updates ---")

# Process every reward
for machine in machines:

    print(f"\nMachine {machine}")

    for reward in rewards[machine]:

        # Increment count
        N[machine] += 1

        # Incremental update formula
        Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

        print(
            f"Reward = {reward}, "
            f"N({machine}) = {N[machine]}, "
            f"Q({machine}) = {Q[machine]:.2f}"
        )

        # Find current greedy machine
        selected = max(Q, key=Q.get)

        print(f"Current Greedy Selection: {selected}")

print("\n--- Final Q Values ---")

for machine in machines:
    print(f"Q({machine}) = {Q[machine]:.2f}")

selected_machine = max(Q, key=Q.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```
