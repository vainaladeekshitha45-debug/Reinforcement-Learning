```python
# Practical 8: State Value and Action Value

# Grid World configuration
gamma = 0.9

rewards = {
    'normal': -1,
    'goal': 10,
    'obstacle': -5
}

# State positions
positions = {
    'S0': (0, 0),
    'S1': (0, 1),
    'G':  (0, 2),
    'X':  (1, 0),
    'S3': (1, 1),
    'S4': (1, 2)
}

# Given policy
policy = {
    'S0': 'Right',
    'S1': 'Down',
    'S3': 'Right',
    'S4': 'Up'
}

# Action movements
actions = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}


# Find next state and reward
def get_next_state(state, action):

    row, col = positions[state]

    dr, dc = actions[action]

    new_row = row + dr
    new_col = col + dc

    # Check grid boundaries
    if not (0 <= new_row < 2 and 0 <= new_col < 3):
        return state, rewards['normal']

    # Find destination state
    next_state = None

    for s, pos in positions.items():
        if pos == (new_row, new_col):
            next_state = s
            break

    # Check obstacle
    if next_state == 'X':
        return next_state, rewards['obstacle']

    # Check goal
    if next_state == 'G':
        return next_state, rewards['goal']

    return next_state, rewards['normal']


# Calculate State Value Function
def state_value_function(state):

    total_return = 0
    discount = 1

    while state != 'G' and state != 'X':

        action = policy[state]

        next_state, reward = get_next_state(state, action)

        total_return += discount * reward

        discount *= gamma

        state = next_state

    return total_return


# Calculate Action Value Function
def action_value_function(state, action):

    total_return = 0
    discount = 1

    # First take the given action
    next_state, reward = get_next_state(state, action)

    total_return += discount * reward

    discount *= gamma

    state = next_state

    # Follow policy for remaining steps
    while state != 'G' and state != 'X':

        action = policy[state]

        next_state, reward = get_next_state(state, action)

        total_return += discount * reward

        discount *= gamma

        state = next_state

    return total_return


# Display results
print("State Value Function")
print("--------------------")

v_s0 = state_value_function('S0')

print("V(S0) =", round(v_s0, 3))


print("\nAction Value Function")
print("---------------------")

q_s0_up = action_value_function('S0', 'Up')

print("Q(S0, Up) =", round(q_s0_up, 3))
```
