```python
# Practical 9 - Assignment 1
# Bellman Expectation Equations
# Stochastic Policy and Deterministic Environment

gamma = 0.9

# Solve simultaneous equations algebraically
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V_S0 = 71.819 / 0.9271

V_S1 = 89.9 + 0.09 * V_S0

# Calculate action values
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

# Display results
print("ASSIGNMENT 1 RESULTS")
print("--------------------")

print("State Values:")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("\nAction Values:")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Verification
print("\nVerification:")
print("V(S0) =", round(
    0.9 * Q_S0_Down + 0.1 * Q_S0_Right, 2
))

print("V(S1) =", round(
    0.9 * Q_S1_Right + 0.1 * Q_S1_Up, 2
))
```

```python
# Practical 9 - Assignment 2
# Iterative Policy Evaluation
# Stochastic Environment and Deterministic Policy

gamma = 0.9
theta = 0.000001

# Initial state values
V_S0 = 0.0
V_S1 = 0.0

# Deterministic policy
policy = {
    'S0': 'Down',
    'S1': 'Right'
}

# Iterative policy evaluation
iterations = 0

while True:

    # Store previous values
    old_S0 = V_S0
    old_S1 = V_S1

    # Calculate new values using old values
    new_S0 = (
        0.1 * (-1 + gamma * 0) +
        0.9 * (-1 + gamma * old_S1)
    )

    new_S1 = (
        0.9 * (100 + gamma * 0) +
        0.1 * (-1 + gamma * old_S0)
    )

    # Calculate maximum difference
    delta = max(
        abs(new_S0 - old_S0),
        abs(new_S1 - old_S1)
    )

    # Update state values
    V_S0 = new_S0
    V_S1 = new_S1

    iterations += 1

    # Check convergence
    if delta < theta:
        break


# Calculate action values after convergence

# S0 - Down: Wall (0.1), S1 (0.9)
Q_S0_Down = (
    0.1 * (-1 + gamma * 0) +
    0.9 * (-1 + gamma * V_S1)
)

# S0 - Right: Wall (1.0)
Q_S0_Right = -1 + gamma * 0

# S1 - Right: Goal (0.9), S0 (0.1)
Q_S1_Right = (
    0.9 * (100 + gamma * 0) +
    0.1 * (-1 + gamma * V_S0)
)

# S1 - Up: S0 (1.0)
Q_S1_Up = -1 + gamma * V_S0


# Display results
print("ASSIGNMENT 2 RESULTS")
print("--------------------")

print("Number of iterations:", iterations)

print("\nState Values:")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("\nAction Values:")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Policy consistency checks
print("\nPolicy Consistency Checks:")

print("V(S0) = Q(S0, Down):",
      round(V_S0, 2) == round(Q_S0_Down, 2))

print("V(S1) = Q(S1, Right):",
      round(V_S1, 2) == round(Q_S1_Right, 2))
```
