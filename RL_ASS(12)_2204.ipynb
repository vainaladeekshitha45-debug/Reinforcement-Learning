```python
# Practical 12
# Value Iteration in Deterministic Grid World

# Parameters
gamma = 0.9
theta = 0.000001

# Define states and actions
states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# Transition mapping
# Format: (Next State, Reward)

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },

    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

# Initialize state values
V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}


# Function to calculate next state value
def get_value(state, values):

    if state == 'G':
        return 0.0

    return values[state]


# Display MDP information
print("GRID WORLD MDP")
print("----------------")

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward: -1")
print("Goal Reward: +100")

print("\nSTATE TRANSITIONS")
print("-----------------")

for state in states:

    for action in actions:

        next_state, reward = transitions[state][action]

        print(
            state, "+", action,
            "->", next_state,
            "Reward =", reward
        )


# Value Iteration
print("\nVALUE ITERATION")
print("----------------")

iteration = 0

print("Iteration | V(S0) | V(S1) | Delta")

while True:

    # Store previous state values
    old_V = V.copy()

    # Create new state values
    new_V = old_V.copy()

    # Calculate maximum action value
    for state in states:

        q_values = {}

        for action in actions:

            next_state, reward = transitions[state][action]

            q = reward + gamma * get_value(
                next_state, old_V
            )

            q_values[action] = q

        # Update state value
        new_V[state] = max(q_values.values())

    # Calculate maximum difference
    delta = max(
        abs(new_V['S0'] - old_V['S0']),
        abs(new_V['S1'] - old_V['S1'])
    )

    # Update values
    V = new_V

    iteration += 1

    print(
        iteration, "|",
        round(V['S0'], 6), "|",
        round(V['S1'], 6), "|",
        round(delta, 6)
    )

    # Check convergence
    if delta < theta:
        break


# Calculate final action values
print("\nFINAL ACTION VALUES")
print("--------------------")

Q = {}

for state in states:

    Q[state] = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        Q[state][action] = (
            reward + gamma * get_value(next_state, V)
        )

        print(
            "Q(", state, ",", action, ") =",
            round(Q[state][action], 2)
        )


# Find optimal policy
print("\nOPTIMAL POLICY")
print("--------------")

policy = {}

for state in states:

    best_action = max(
        actions,
        key=lambda action: Q[state][action]
    )

    policy[state] = best_action

    print(state, ":", best_action)


# Display optimal state values
print("\nOPTIMAL STATE VALUES")
print("--------------------")

for state in states:
    print("V*(", state, ") =", round(V[state], 2))

print("V*(G) = 0")


# Bellman Optimality verification
print("\nBELLMAN OPTIMALITY VERIFICATION")
print("--------------------------------")

for state in states:

    max_q = max(Q[state].values())

    print(
        state,
        "V* =", round(V[state], 2),
        "Max Q* =", round(max_q, 2),
        "Verified =", abs(V[state] - max_q) < 1e-9
    )


# Final result
print("\nFINAL RESULT")
print("------------")

print("Total Iterations:", iteration)
print("Optimal Path: S0 -> S1 -> G")
print("Optimal Policy:", policy)
```
