```python
# Practical 11
# Policy Iteration in Deterministic Grid World

# Parameters
gamma = 0.9
theta = 0.000001

# States and actions
states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# Transition mapping
# Format: (Next State, Reward)

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },

    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

# Terminal state
terminal_value = 0

# Initial policy
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

# Initialize state values
V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}


# Get next state value
def get_value(state, values):

    if state == 'G':
        return terminal_value

    return values[state]


# Display MDP
print("GRID WORLD MDP")
print("----------------")

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward: -1")
print("Goal Reward: +100")

print("\nTRANSITIONS")
print("-----------")

for state in states:
    for action in actions:

        next_state, reward = transitions[state][action]

        print(state, action, "->",
              next_state, "Reward:", reward)


# Policy Iteration
policy_iteration = 0

while True:

    policy_iteration += 1

    print("\nPOLICY ITERATION:", policy_iteration)
    print("Current Policy:", policy)

    # PART 1: Policy Evaluation
    print("\nPolicy Evaluation")

    evaluation_iteration = 0

    while True:

        old_V = V.copy()
        new_V = old_V.copy()

        for state in states:

            action = policy[state]

            next_state, reward = transitions[state][action]

            new_V[state] = (
                reward + gamma * get_value(next_state, old_V)
            )

        delta = max(
            abs(new_V['S0'] - old_V['S0']),
            abs(new_V['S1'] - old_V['S1'])
        )

        V = new_V
        evaluation_iteration += 1

        print(
            "Iteration:", evaluation_iteration,
            "V(S0):", round(V['S0'], 6),
            "V(S1):", round(V['S1'], 6),
            "Delta:", round(delta, 8)
        )

        if delta < theta:
            break

    # PART 2: Calculate action values
    Q = {}

    for state in states:

        Q[state] = {}

        for action in actions:

            next_state, reward = transitions[state][action]

            Q[state][action] = (
                reward + gamma * get_value(next_state, V)
            )

    # PART 3: Policy Improvement
    print("\nPolicy Improvement")

    old_policy = policy.copy()

    for state in states:

        best_action = max(
            actions,
            key=lambda action: Q[state][action]
        )

        policy[state] = best_action

    print("Updated Policy:", policy)

    # PART 4: Check policy stability
    if policy == old_policy:

        print("\nPolicy is stable!")
        break


# Final Results
print("\nFINAL OPTIMAL STATE VALUES")
print("--------------------------")

for state in states:
    print(state, "=", round(V[state], 2))

print("G = 0.0")


# Display action values
print("\nFINAL ACTION VALUES")
print("--------------------")

for state in states:

    for action in actions:

        print(
            "Q(", state, ",", action, ") =",
            round(Q[state][action], 2)
        )


# Display optimal policy
print("\nOPTIMAL POLICY")
print("--------------")

for state in states:
    print(state, ":", policy[state])


# Bellman Optimality Verification
print("\nBELLMAN OPTIMALITY VERIFICATION")
print("--------------------------------")

for state in states:

    max_q = max(Q[state].values())

    print(
        state,
        "V =", round(V[state], 2),
        "Max Q =", round(max_q, 2),
        "Verified =", abs(V[state] - max_q) < 1e-9
    )


print("\nFINAL RESULT")
print("------------")
print("Total Policy Iterations:", policy_iteration)
print("Optimal Path: S0 -> S1 -> G")
```
